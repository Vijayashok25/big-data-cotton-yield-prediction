# Big Data Cotton Yield Prediction: Exploratory Data Analysis (EDA)
This notebook conducts in-depth statistical profiling, feature correlation analysis, and data quality audits on the `dhyann2815/india-crop-yield-prediction` dataset.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
data_path = "../data/crop_yield.csv"
df = pd.read_csv(data_path)
print(f"Loaded raw dataset with shape: {df.shape}")
df.head()
```

## 1. Filter Cotton(lint) Records
We isolate cotton crop observations and inspect state distribution.

```python
cotton_df = df[df["Crop"].str.contains("cotton", case=False, na=False)].copy()
print(f"Total Cotton records: {len(cotton_df)}")
print(cotton_df.info())
cotton_df.describe()
```

## 2. Target Variable (Yield) Distribution & Outlier Analysis

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(cotton_df["Yield"], kde=True, ax=ax[0], color="#1e40af")
ax[0].set_title("Distribution of Cotton Yield (tonnes/ha)")
sns.boxplot(x=cotton_df["Yield"], ax=ax[1], color="#93c5fd")
ax[1].set_title("Boxplot of Cotton Yield")
plt.show()
```

## 3. Correlation Heatmap & Input Intensity Ratios
We engineer Fertilizer/Area and Pesticide/Area intensity metrics and plot the correlation matrix.

```python
cotton_df["Fertilizer_per_Area"] = cotton_df["Fertilizer"] / (cotton_df["Area"] + 1e-5)
cotton_df["Pesticide_per_Area"] = cotton_df["Pesticide"] / (cotton_df["Area"] + 1e-5)

num_cols = ["Year", "Area", "Annual_Rainfall", "Fertilizer", "Pesticide", "Fertilizer_per_Area", "Pesticide_per_Area", "Yield"]
plt.figure(figsize=(10, 8))
sns.heatmap(cotton_df[num_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", square=True)
plt.title("Correlation Matrix (Cotton Dataset)")
plt.show()
```

## 4. State-Wise Productivity Ranking

```python
state_agg = cotton_df.groupby("State")["Yield"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 5))
state_agg.plot(kind="bar", color="#059669")
plt.title("Top 10 States by Average Cotton Yield")
plt.ylabel("Mean Yield (tonnes/ha)")
plt.xticks(rotation=45, ha="right")
plt.show()
```
